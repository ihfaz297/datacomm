Write a Python program that reads a WAV file and calculates its transmission time based on the file size and specified data rate?

```python
import os


def transmission_time(wav_file, data_rate):
    """
    Calculate the transmission time of a WAV file.

    Parameters:
        wav_file (str): Path to the WAV file.
        data_rate (float): Data rate in bits per second (bps).

    Returns:
        float: Transmission time in seconds.
    """

    if not os.path.isfile(wav_file):
        raise FileNotFoundError(f"File not found: {wav_file}")

    if data_rate <= 0:
        raise ValueError("Data rate must be greater than 0.")

    # Get file size in bytes
    file_size_bytes = os.path.getsize(wav_file)

    # Convert bytes to bits
    message_size_bits = file_size_bytes * 8

    # Calculate transmission time
    time_seconds = message_size_bits / data_rate

    return time_seconds


# Example
wav_file = "/content/drive/My Drive/Commn Engg/sample-12s.wav"
data_rate = 128000  # 128 kbps

try:
    file_size = os.path.getsize(wav_file)
    time = transmission_time(wav_file, data_rate)

    print(f"Message size: {file_size:,} bytes")
    print(f"Message size: {file_size * 8:,} bits")
    print(f"Data rate: {data_rate:,} bps")
    print(f"Transmission time: {time:.4f} seconds")

except (FileNotFoundError, ValueError) as e:
    print(f"Error: {e}")
```

A network-layer packet contains 1,000 bytes of data.
At the Data Link Layer, the Ethernet frame contains:

•	Destination MAC address = 6 bytes

•	Source MAC address = 6 bytes

•	Type/Length field = 2 bytes

•	Payload = 1,000 bytes

•	FCS = 4 bytes

Write a Python program to calculate:
1.	Ethernet frame size.
2.	Total overhead.
3.	Encapsulation efficiency.


```python
payload = 1000
destination_mac = 6
source_mac = 6
type_field = 2
fcs = 4

header = destination_mac + source_mac + type_field
trailer = fcs

frame_size = header + payload + trailer
overhead = header + trailer

efficiency = (payload / frame_size) * 100

print("\n--- Ethernet Frame ---")
print(f"Payload:          {payload} bytes")
print(f"Header:           {header} bytes")
print(f"Trailer (FCS):    {trailer} bytes")
print(f"Total frame size: {frame_size} bytes")
print(f"Total overhead:   {overhead} bytes")
print(f"Efficiency:       {efficiency:.2f}%")
```

A user sends 500 bytes of application data.

The following headers are added:

Transport Layer: 20 bytes

Network Layer: 20 bytes

Data Link Layer header: 14 bytes

Data Link Layer trailer: 4 bytes

Write a Python program that displays the size of the data after each OSI encapsulation step.

```python
application_data = 500

transport_header = 20
network_header = 20
data_link_header = 14
data_link_trailer = 4

transport_data = application_data + transport_header
network_packet = transport_data + network_header
data_link_frame = (
    network_packet
    + data_link_header
    + data_link_trailer
)

print("--- OSI Encapsulation ---")

print(f"Application Layer : {application_data} bytes")
print(f"Transport Layer   : {transport_data} bytes")
print(f"Network Layer     : {network_packet} bytes")
print(f"Data Link Layer   : {data_link_frame} bytes")

print("\nTotal overhead:",
      data_link_frame - application_data,
      "bytes")
efficiency = (application_data / data_link_frame) * 100
print(f"Efficiency:       {efficiency:.2f}%")
```

Write a Python program to:

Perform bit stuffing.

Display the original data.

Display the stuffed data.

Calculate the number of inserted bits.

```python
def bit_stuff(data):
    stuffed = ""
    consecutive_ones = 0
    inserted_bits = 0

    for bit in data:
        stuffed += bit

        if bit == "1":
            consecutive_ones += 1

            if consecutive_ones == 5:
                stuffed += "0"
                inserted_bits += 1
                consecutive_ones = 0
        else:
            consecutive_ones = 0

    return stuffed, inserted_bits


data = input("Enter binary data: ")

stuffed_data, inserted = bit_stuff(data)

print("\n--- Bit Stuffing ---")
print("Original data:", data)
print("Stuffed data: ", stuffed_data)
print("Original size:", len(data), "bits")
print("Stuffed size: ", len(stuffed_data), "bits")
print("Inserted bits:", inserted)
```
